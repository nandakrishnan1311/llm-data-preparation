```python
import re

def clean_text(text):
    if not text:
        return ""

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove Wikipedia-style citation markers such as [1], [23]
    text = re.sub(r'\[\d+\]', '', text)

    # Remove unwanted control characters
    text = re.sub(
        r'[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]',
        '',
        text
    )

    # Replace multiple whitespace characters with one space
    text = re.sub(r'\s+', ' ', text)

    # Remove leading/trailing whitespace
    text = text.strip()

    return text


# Test
cleaned = clean_text(english[0]["text"])

print(cleaned[:2000])
```

```python
import re
import unicodedata

# Wikipedia sections that usually contain references/navigation
STOP_SECTIONS = {
    # English
    "references",
    "external links",
    "see also",
    "further reading",
    "bibliography",
    "notes",
    "sources",

    # Hindi
    "सन्दर्भ",
    "संदर्भ",
    "बाहरी कड़ियाँ",
    "बाहरी कड़ियां",
    "यह भी देखें",
    "आगे पढ़ें",
    "स्रोत"
}

def clean_text(text):
    if not text:
        return ""

    # Normalize Unicode
    text = unicodedata.normalize("NFKC", text)

    # Normalize special spaces
    text = text.replace("\xa0", " ")
    text = text.replace("\u200b", "")
    text = text.replace("\ufeff", "")

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove Wikipedia-style numeric citations: [1], [23], etc.
    text = re.sub(r'\[\s*\d+\s*\]', '', text)

    # Remove common citation markers
    text = re.sub(
        r'\[\s*(?:citation needed|not verified|failed verification)\s*\]',
        '',
        text,
        flags=re.IGNORECASE
    )

    # Remove HTML/XML tags if present
    text = re.sub(r'<[^>]+>', '', text)

    # Remove unwanted control characters, but keep \n and \t
    text = re.sub(r'[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]', '', text)

    # Process paragraph by paragraph
    paragraphs = []

    for paragraph in text.split("\n"):
        paragraph = re.sub(r'[ \t]+', ' ', paragraph).strip()

        if not paragraph:
            continue

        # Check whether this is a Wikipedia section we want to stop at
        heading = paragraph.strip().lower()

        if heading in STOP_SECTIONS:
            break

        paragraphs.append(paragraph)

    # Preserve paragraph boundaries
    text = "\n\n".join(paragraphs)

    return text.strip()
```

```python
TARGET_MB = 150
TARGET_BYTES = TARGET_MB * 1024 * 1024


def collect_articles(dataset, target_bytes, language):
    collected = []
    total_bytes = 0

    for item in dataset:
        text = item["text"]

        # Skip empty text
        if not text or not text.strip():
            continue

        text_bytes = len(text.encode("utf-8"))

        collected.append({
            "id": item["id"],
            "title": item["title"],
            "url": item["url"],
            "text": text,
            "language": language
        })

        total_bytes += text_bytes

        if total_bytes >= target_bytes:
            break

    print(f"{language} articles collected:", len(collected))
    print(
        f"{language} raw size:",
        round(total_bytes / (1024 * 1024), 2),
        "MB"
    )

    return collected
```

```python
def remove_duplicates_with_stats(data):
    seen = set()
    unique_data = []
    duplicates = 0

    for item in data:
        text = item["text"].strip()

        if text in seen:
            duplicates += 1
        else:
            seen.add(text)
            unique_data.append(item)

    return unique_data, duplicates


unique_english_full, english_duplicates = remove_duplicates_with_stats(
    cleaned_english_full
)

unique_hindi_full, hindi_duplicates = remove_duplicates_with_stats(
    cleaned_hindi_full
)

print("English")
print("Before:", len(cleaned_english_full))
print("After:", len(unique_english_full))
print("Duplicates removed:", english_duplicates)

print("\nHindi")
print("Before:", len(cleaned_hindi_full))
print("After:", len(unique_hindi_full))
print("Duplicates removed:", hindi_duplicates)
```

```python
from langdetect import detect, DetectorFactory

# Make language detection reproducible
DetectorFactory.seed = 0

def check_languages(data, expected_language, sample_size=100):
    correct = 0
    incorrect = 0
    errors = 0

    sample = data[:sample_size]

    for item in sample:
        try:
            detected = detect(item["text"])

            if detected == expected_language:
                correct += 1
            else:
                incorrect += 1

        except Exception:
            errors += 1

    print("Expected language:", expected_language)
    print("Sample size:", len(sample))
    print("Correct:", correct)
    print("Incorrect:", incorrect)
    print("Detection errors:", errors)


print("ENGLISH CHECK")
check_languages(
    unique_english_full,
    "en"
)

print("\nHINDI CHECK")
check_languages(
    unique_hindi_full,
    "hi"
)
```

```python
# Full English tokenization in batches

BATCH_SIZE = 100

english_total_tokens = 0

for start in range(0, len(unique_english_full), BATCH_SIZE):
    batch = unique_english_full[start:start + BATCH_SIZE]

    for item in batch:
        tokens = tokenizer.encode(
            item["text"],
            add_special_tokens=False
        )
        english_total_tokens += len(tokens)

    if (start + BATCH_SIZE) % 1000 == 0:
        print(f"Processed {min(start + BATCH_SIZE, len(unique_english_full))} English articles")

print("\nEnglish tokenization complete")
print("Articles:", len(unique_english_full))
print("Total tokens:", english_total_tokens)
print(
    "Average tokens/article:",
    english_total_tokens / len(unique_english_full)
)
```
